```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
df = pd.read_csv(root / "data/raw/AbandonoEscolar_RendaMedia_2013_2023.csv")
display(df.head(), df.isna().sum())
```

```python
base = df.query("Unidade_Geografica == 'Brasil' and Localizacao == 'Total' and Dependencia_Administrativa == 'Total' and Grupo_de_Abandono in ['Ensino Fundamental', 'Ensino Médio']")
pivo = base.pivot(index="Ano", columns="Grupo_de_Abandono", values="Taxa_Abandono")
display(pivo)
pivo.plot(marker="o", title="Abandono escolar nacional (%)")
plt.ylabel("Taxa (%)")
plt.show()
```

```python
for etapa, grupo in base.groupby("Grupo_de_Abandono"):
    print(etapa, "Pearson:", grupo["Taxa_Abandono"].corr(grupo["Renda_Media"]))
```
